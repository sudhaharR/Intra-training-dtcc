## Question 4 Revenue by city

```python
df["Revenue"] = df["Price"] * df["Sales_Million"]
```

```python
genre_revenue = df.groupby("Genre")["Revenue"].sum()

print(genre_revenue.sort_values(ascending=False))
```

```python
highest_genre = genre_revenue.idxmax()
highest_revenue = genre_revenue.max()

print("Highest Revenue Genre:", highest_genre)
print("Revenue:", highest_revenue)
```

```python
platform_revenue = df.groupby("Platform")["Revenue"].sum()

print(platform_revenue.sort_values(ascending=False))
```

```python
print("Highest Revenue Platform:",
      platform_revenue.idxmax())
```

```python
def popularity(revenue):
    if revenue < 1000:
        return "Low"
    elif revenue <= 3000:
        return "Medium"
    else:
        return "High"

df["Popularity"] = df["Revenue"].apply(popularity)
```

```python
print(df["Popularity"].value_counts())
```

```python
df["Popularity"].value_counts().plot(kind="bar")

plt.title("Game Popularity Distribution")
plt.xlabel("Popularity")
plt.ylabel("Number of Games")
plt.show()
```

```python
popular_games = df[df["Popularity"] == "High"]

print(
    popular_games[
        ["Game_Name", "Revenue", "Popularity"]
    ]
)
```
