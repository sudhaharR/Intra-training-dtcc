```python
df[df["Sales_Million"] < 0]
```

## Detecting outliers

```python
Q1 = df["Sales_Million"].quantile(0.25)
```

```python
Q3 = df["Sales_Million"].quantile(0.75)
```

```python
IQR = Q3 - Q1
```

```python
lower = Q1 - 1.5 * IQR
```

```python
upper = Q3 + 1.5 * IQR
```

```python
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower:", lower)
print("Upper:", upper)
```

```python
outliers = df[
    (df["Sales_Million"] < lower) |
    (df["Sales_Million"] > upper)
]

print(outliers)
```

```python
import matplotlib.pyplot as plt

plt.boxplot(df["Sales_Million"])
plt.title("Sales Million - Outlier Detection")
plt.ylabel("Sales")
plt.show()
```

## Feature Engineering

```python
df["Revenue"] = df["Price"] * df["Sales_Million"]
```

```python
df["Revenue"]
```

```python
def popularity(revenue):
    if revenue < 1000:
        return "Low"
    elif revenue <= 3000:
        return "Medium"
    else:
        return "High"
```

```python
df["Popularity"] = df["Revenue"].apply(popularity)
```

```python
print(df[["Game_Name", "Price", "Sales_Million", "Revenue", "Popularity"]])
```
